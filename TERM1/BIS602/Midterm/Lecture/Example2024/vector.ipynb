```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
# Initialize sentence embedding model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
# Sample sentences
sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "A fast, dark fox leaps over a sleeping canine.",
    "The sky is blue and the sun is shining."
]

# Convert sentences to embeddings
embeddings = model.encode(sentences)
```

```python
# Create FAISS index
dimension = embeddings.shape[1]  # Embedding size
index = faiss.IndexFlatL2(dimension)  # L2 (Euclidean) distance
index.add(embeddings)  # Store embeddings in FAISS
```

```python
#Retrieve stored vectors
stored_vectors = index.reconstruct_n(0, len(sentences))  # Get all stored vectors
stored_vectors = np.array(stored_vectors)  # Convert to NumPy array
print(f"Vector of the fist sentence: '{stored_vectors[0]}")
```

```python
# Query sentence
query_sentence = "A fast fox jumps over a lazy dog."
query_embedding = model.encode([query_sentence])

# Search FAISS for similar sentences
top_k = 3  # Number of closest matches
distances, indices = index.search(query_embedding, top_k)
```

```python
# Display results
print(f"\nQuery: {query_sentence}")
for i, idx in enumerate(indices[0]):
    print(f"Matched Sentence: {sentences[idx]} (Distance: {distances[0][i]:.4f})")
```
