```python
import pandas as pd
```

```python
# **Extract**: Read the CSV file
def extract(file_path):
    return pd.read_csv(file_path)
```

```python
# **Transform**: Clean and modify data
def transform(df):
    df.fillna({'age': df['age'].mean(), 'score': df['score'].mean()}, inplace=True)  # Fill missing values
    df['grade'] = df['score'].apply(lambda x: 'A' if x >= 85 else 'B' if x >= 75 else 'C')  # Add a grade column
    return df
```

```python
# **Load**: Save to a new CSV file
def load(df, output_path):
    df.to_csv(output_path, index=False)
    print(f"Data loaded successfully into {output_path}")
```

```python
# Run the ETL process
if __name__ == "__main__":
    input_file = "/content/sample_data/data.csv"
    output_file = "/content/sample_data/cleaned_data.csv"

    # ETL execution
    data = extract(input_file)
    transformed_data = transform(data)
    load(transformed_data, output_file)

    print(transformed_data)
    print("ETL process completed!")
```
