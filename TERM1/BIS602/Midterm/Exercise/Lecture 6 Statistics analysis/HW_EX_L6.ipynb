Task 1: Load 3 Months of Data for Two Thai Stocks, e.g. PTT PCL. (PTT.BK) and Bangchak Corporation (BCP.BK), both key players in Thailand’s energy sector.
Use yfinance to download daily stock prices for the past 3 months.

```python
import yfinance as yf
import pandas as pd
import numpy as np

# Define stock tickers
stocks = ["PTT.BK", "BCP.BK"]

# Download 3 months of daily data
df = yf.download(stocks, period="3mo")

# Display first few rows
df.head()
```

```python
print(df.columns)
```

```python
df1 = df['Close']
df1.head()
```

Task 2: Compute Daily Returns
Calculate daily percentage returns for both stocks.

```python
# Compute daily returns
returns = df1.pct_change().dropna()

# Display first few rows
returns.head()
```
