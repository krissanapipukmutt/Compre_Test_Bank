Use yfinance to download daily stock prices for the past 3 months.

```python
import yfinance as yf
import pandas as pd
import numpy as np

# Define stock tickers
stocks = ["KBANK.BK", "BAY.BK"]

# Download 3 months of daily data
df = yf.download(stocks, period="3mo")

# Display first few rows
df
```

Task 3: Compute Key Metrics

3.1 Coefficient of Variation (CV): Measures risk per unit of return.

3.2 Sharpe Ratio: Measures return per unit of risk, Assume risk-free rate 1.5% per year (~0.00006 daily).

```python
#3.1 Coefficient of Variation (CV): Measures risk per unit of return.
sample_std_dev_KBANK = np.std(df1[('KBANK.BK')], ddof=1) # Access column using MultiIndex
sample_std_dev_BAY = np.std(df1[('BAY.BK')], ddof=1)  # Access column using MultiIndex

print(f"\nStandard Deviation_KBANK: {sample_std_dev_KBANK:.6f}")
print(f"\nmean_KBANK: {df1[('KBANK.BK')].mean():.6f}")

print(f"\nStandard Deviation_BAY: {sample_std_dev_BAY:.6f}")
print(f"\nmean_BAY: {df1[('BAY.BK')].mean():.6f}")


coeff_variation_KBANK = (sample_std_dev_KBANK / df1[('KBANK.BK')].mean()) # Access column using MultiIndex
print(f"\nCoefficient of Variation for KBANK: {coeff_variation_KBANK:.6f}%")

coeff_variation_BAY = (sample_std_dev_BAY / df1[('BAY.BK')].mean()) # Access column using MultiIndex
print(f"\nCoefficient of Variation for BEY: {coeff_variation_BAY:.6f}%")
```

```python
#3.2 Sharpe Ratio: Measures return per unit of risk, Assume risk-free rate 1.5% per year (~0.00006 daily).
# Assuming annual returns, calculate Sharpe Ratio
risk_free_rate = 0.00006  # 0.00006% risk-free rate

# Calculate the excess returns
excess_returns_KBANK = df1[('KBANK.BK')] - risk_free_rate
excess_returns_BEY = df1[('BAY.BK')] - risk_free_rate

# Calculate the Sharpe Ratio
sharpe_ratio_KBANK = excess_returns_KBANK.mean() / excess_returns_KBANK.std()
sharpe_ratio_BEY = excess_returns_BEY.mean() / excess_returns_BEY.std()

print(f"Sharpe Ratio for KBANK: {sharpe_ratio_KBANK:.2f}")
print(f"Sharpe Ratio for BEY: {sharpe_ratio_BEY:.2f}")
```

Task 4: Compute the Shape of distibution for the two stocks:
*   Skewness: Measures return distribution symmetry.
*   Kurtosis: Measures tail risk (excess kurtosis).


```python
#Task 4: Compute the Shape of distibution for the two stocks:
#Skewness: Measures return distribution symmetry.
kewness_KBANK = df1[('KBANK.BK')].skew()
print(f"\nSkewness for KBANK: {kewness_KBANK:.2f}")

skewness_BAY = df1[('BAY.BK')].skew()
print(f"\nSkewness for BAY: {skewness_BAY:.2f}")

#Kurtosis: Measures tail risk (excess kurtosis).
kurtosis_KBANK = df1[('KBANK.BK')].kurtosis()
print(f"\nKurtosis for KBANK: {kurtosis_KBANK:.2f}")

kurtosis_BAY = df1[('BAY.BK')].kurtosis() # Use the correct ticker 'BAY.BK'
print(f"\nKurtosis for BEY: {kurtosis_BAY:.2f}")
#Beware that Kurtosis in Python is calculated by a reference of zero for Normal dist, so Kurtosis > 0 is Lepto (higher peak)
```
