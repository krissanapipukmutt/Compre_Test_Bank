```python
rng = np.random.RandomState(2021)
```

```python
ST_Percent_Return = rng.normal(2,0.3,100)
MT_Percent_Return = rng.normal(5,2.5,100)
LT_Percent_Return = rng.normal(6.25,5.5,100)
```

```python
Invest = 300000
Port = [Invest*.45, Invest*.35, Invest*.20]
```

```python
Port
```

```python
df=[]
```

```python
df = pd.DataFrame(list(zip(ST_Percent_Return, MT_Percent_Return, LT_Percent_Return)), columns=['ST_%return', 'MT_%return', 'LT_%return'])
```

```python
df['ST_Return'] = ST_Percent_Return*Port[0]/100
```

```python
df
```

```python
df['MT_Return'] = MT_Percent_Return*Port[1]/100
df['LT_Return'] = LT_Percent_Return*Port[2]/100
```

```python
df
```

```python
df['Total Return']=(df.ST_Return+df.MT_Return+df.LT_Return)
```

```python
df
```

```python
df['Ending Balance']=df['Total Return']+Invest
```

```python
df
```

```python
Balance_mean = np.mean(df['Ending Balance'])
Balance_stdev = np.std(df['Ending Balance'])
```

```python
print(Balance_mean, Balance_stdev)
```
