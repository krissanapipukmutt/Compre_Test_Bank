```python
import numpy as np

# Define the values and frequencies
values = [100, 200, 300, 400]
frequencies = [20, 25, 35, 10]

# Normalize frequencies to probabilities
probabilities = np.array(frequencies) / np.sum(frequencies)

# Generate random samples based on the discrete distribution
samples = np.random.choice(values, size=100, p=probabilities)

# Display the first few samples
print(samples[:10])
```

```python
mean = np.mean(samples)
stdev = np.std(samples)
```

```python
print('mean =', mean ,'\n', 'stdev = ', stdev)
```
