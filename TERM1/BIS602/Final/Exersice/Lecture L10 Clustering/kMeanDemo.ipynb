```python
import pandas as pd
```

```python
#!pip install scikit-learn
import sklearn
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
data_read = pd.read_csv (r'/content/sample_data/clustering_subjects.csv')
data = data_read.iloc[:,:3].values
```

```python
data
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html
Sum_of_squared_distances = []
K = range(1,6)
for num_clusters in K :
 kmeans = KMeans(n_clusters=num_clusters,n_init=6)
 kmeans.fit(data)
 Sum_of_squared_distances.append(kmeans.inertia_)

print(f"Inertias of K = 1 to 5:  {Sum_of_squared_distances}")
```

```python
plt.plot(K,Sum_of_squared_distances,'bx-')
plt.xlabel('Values of K')
plt.ylabel('Sum of squared distances/Inertia')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
# n_int is to set max no. of clusters
cluster = KMeans(n_clusters=3, n_init=6).fit(data)
print(cluster.cluster_centers_)
print(cluster.labels_)
```

```python
import numpy as np
from matplotlib.colors import ListedColormap
#colormap = np.array(['r', 'g', 'b'])
colours = ListedColormap(['r','b','g'])
#scatter=plt.scatter(data[:,0],data[:,1], c=colormap[cluster.labels_], label=colormap)
scatter=plt.scatter(data[:,0],data[:,1], c=cluster.labels_, cmap=colours)
plt.legend(*scatter.legend_elements(),loc="lower right")

plt.show()
```
