```python
import duckdb

conn = duckdb.connect("my_database.db")
print("✅ Connected to DuckDB file my_database.db!")
```

```python
# Load a CSV file into DuckDB
#https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv
conn.execute("""
    CREATE OR REPLACE TABLE sales_data AS
    SELECT * FROM read_csv_auto('/content/OLAP_Data.csv')
""")

print("✅ CSV Data Loaded into DuckDB!")
```

```python
# Query to preview the first 5 rows of the sales_data table
preview_query = "SELECT * FROM sales_data LIMIT 5"

df_preview = conn.execute(preview_query).fetchdf()
display(df_preview)
```

```python
#Query using SQL
query = """
SELECT
    Year, Product, Region, SUM(Sales_Amount) AS Total_Sales
FROM sales_data
WHERE Region IN ('USA', 'UK') AND Year >= 2023
GROUP by Year, Region, Product
ORDER BY Total_Sales DESC
"""

df_result = conn.execute(query).fetchdf()
df_result
```

```python
conn.close()
print("✅ DuckDB connection closed!")
```
