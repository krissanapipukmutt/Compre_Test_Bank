# **Section 1:** Bar, Stack bar, Histogram, and basic Scartter plot




```python
import pandas as pd
df = pd.read_excel('/content/sample_data/Book L9 Visualiz.xlsx', sheet_name='Construction clothing')
df.head()
```

```python
#EX 1: Plot bar graph for frequency of Repeat column
import matplotlib.pyplot as plt
df['Repeat'].value_counts().plot(kind='bar')
plt.xlabel("Repeat Values")
plt.ylabel("Frequency")
plt.title("Frequency of Repeat Values")
plt.show()
```

```python
#EX 2: Plot histogram of the income column, break values into 5 bins equally.
#EX 2: Plot histogram of the income column, break values into 5 bins equally.

import matplotlib.pyplot as plt
plt.hist(df['Income'], bins=5, range=(0, 250), bottom=0)
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Histogram of Income")
plt.show()
```

```python
#EX 3: scatter plot, x = Income and y=Purchase column.

import matplotlib.pyplot as plt
#EX 3: Scatter plot, x = Income and y=Purchase column.
plt.scatter(df['Income'], df['Purchase'])
plt.xlabel("Income")
plt.ylabel("Purchase")
plt.title("Scatter Plot of Income vs Purchase")
plt.show()
```

```python
df = pd.read_excel('/content/sample_data/Book L9 Visualiz.xlsx', sheet_name='Promotion')
df.head()
```

```python
# prompt: Plot stack bars for Sum the number of each Purchase value grouped by Location column

# Group by 'Location' and 'Purchase' to count occurrences
purchase_by_location = df.groupby(['Location', 'Purchase'])['Purchase'].count().unstack()

# Create the stacked bar plot
ax = purchase_by_location.plot(kind='bar', stacked=True)

# Add data labels inside the bars
for i, bar in enumerate(ax.patches):
    height = bar.get_height()
    width = bar.get_width()
    x = bar.get_x() + width / 2
    y = bar.get_y() + height / 2
    label_text = int(height)  # Get the count value as an integer
    ax.text(x, y, label_text, ha='center', va='center', color='white')

# Customize the plot
plt.xlabel("Location")
plt.ylabel("Count of Purchase")
plt.title("Count of Purchase by Location")
plt.show()
```

## Section 2: Scatter and Bubble plot



```python
df = pd.read_excel('/content/sample_data/Book L9 Visualiz.xlsx', sheet_name='Birth_Life')
df.head()
```

```python
#Scatter plot Life_Exp as X and Birth_Rate as Y, having difference color for each value of Development

for development_value in df['Development'].unique():
    subset = df[df['Development'] == development_value]
    plt.scatter(subset['Life_Exp'], subset['Birth_Rate'], label=development_value)

plt.xlabel("Life Expectancy")
plt.ylabel("Birth Rate")
plt.title("Life Expectancy vs. Birth Rate by Development")
plt.legend()  # Show the legend
plt.show()
```

```python
# prompt: Bubble plot Life_Exp as X and Birth_Rate as Y, having difference size by GNI value.

plt.figure(figsize=(10, 6))  # Adjust figure size for better visualization

for i in range(len(df)):
    plt.scatter(df['Life_Exp'][i], df['Birth_Rate'][i], s=df['GNI'][i]/50, alpha=0.7) # Adjust size scaling as needed

plt.xlabel("Life Expectancy")
plt.ylabel("Birth Rate")
plt.title("Life Expectancy vs. Birth Rate (Bubble Plot by GNI)")
plt.show()
```

# **Section 3:** Line plot

```python
df = pd.read_excel('/content/sample_data/Book L9 Visualiz.xlsx', sheet_name='Apple_Merck')
df.head()
```

```python
#Line plot for Apple and Merck, X-axis using Date.

# Convert 'Date' column to datetime objects
df['Date'] = pd.to_datetime(df['Date'])

# Create the line plot
plt.figure(figsize=(10, 6))
plt.plot(df['Date'], df['Apple '], label='Apple')
plt.plot(df['Date'], df['Merck'], label='Merck')

# Customize the plot
plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.title("Apple and Merck Stock Prices Over Time")
plt.legend()
plt.grid(True)  # Add a grid for better readability
plt.xticks(rotation=45) # Rotate x-axis labels for better visibility if needed
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```
