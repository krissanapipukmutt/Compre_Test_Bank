```python
import pandas as pd
import pandas as pd
#df = pd.read_excel(file_path, sheet_name=sheet_name)
df = pd.read_excel("/content/sample_data/Book L8 data treatment and transform st.xlsx", "Restaurant_Reviews")
print(f"shape {df.shape} .... \n{df.head()}")
```

```python
nan_counts = df.isna().sum()
print(f"NaN counts per column:\n{nan_counts}")

nan_positions = df[df.isna().any(axis=1)] #axis 0 rows, 1 columns
print(f"\nRows with NaN values:\n{nan_positions}")
```

```python
df_new = df.copy() # To reserve no change in the orgininal versionf of file
```

```python
# Omission, Delete all rows with NaN

df_cleaned = df.dropna()
print(f"Shape after dropping NaN rows: {df_cleaned.shape}")
nan_positions = df_cleaned[df.isna().any(axis=1)] #axis 0 rows, 1 columns
print(f"\nRows with NaN values:\n{df_cleaned[df.isna().any(axis=1)]}")
```

```python
#Impute NaN by average value

# Calculate the mean of each column, ignoring NaN values
column_means = df.mean(numeric_only=True).round(1)
print(f"Column means {column_means}")
# Fill NaN values with the mean of their respective columns
df_imputed = df.fillna(column_means)

# Verify that there are no more NaN values in the numeric columns
nan_counts_imputed = df_imputed.isna().sum()
print(f"\nNaN counts per column after imputation:\n{nan_counts_imputed}")

# Show all positions imputed
nan_rows_index = df[df.isna().any(axis=1)].index
print(f"\nrows with replacing NaN values:\n{df_imputed.loc[nan_rows_index]}")
```
